```python
import joblib

preprocessor = joblib.load("../models/preprocessor.pkl")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
X_train=pd.read_csv("../data/processed-data/x_train_data.csv")
X_test=pd.read_csv("../data/processed-data/x_test_data.csv")
y_train=pd.read_csv("../data/processed-data/y_train_data.csv")
y_test=pd.read_csv("../data/processed-data/y_test_data.csv")
```

```python
model_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

model_pipeline.fit(X_train, y_train)
y_pred=model_pipeline.predict(X_test)
```

```python
y_pred[:15]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, zero_division=0))
print("Recall:", recall_score(y_test, y_pred, zero_division=0))
print("F1 Score:", f1_score(y_test, y_pred, zero_division=0))

print(classification_report(y_test, y_pred, zero_division=0))
```

```python
from sklearn.ensemble import RandomForestClassifier


rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
from sklearn.metrics import (
    accuracy_score
)

print("Accuracy:", accuracy_score(y_test,rf_pred))

```

```python
#principle component analysis
from sklearn.decomposition import PCA

pca_model = Pipeline([
    ("preprocessor", preprocessor),
    ("pca", PCA(n_components=0.95)),
    ("classifier", LogisticRegression(max_iter=1000))
])
pca_model.fit(X_train, y_train)
# Train
pca_model.fit(X_train, y_train)

# Predict
y_pred_custom = pca_model.predict(X_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred_custom))
print("Precision:", precision_score(y_test, y_pred_custom))
print("Recall:", recall_score(y_test, y_pred_custom))
print("F1 score:", f1_score(y_test, y_pred_custom))
```
